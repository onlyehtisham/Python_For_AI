# Part A — Tuples

A tuple is an **ordered, immutable collection**.

```python
student = ("Ali", 20, 3.45)
```

Tuples normally use parentheses `()`. 

````python
student = ("Ali", 20, 3.45)

print(student)
print(type(student))
````

````python
numbers = (10, 20, 30, 40)
students = ("Ali", "Ahmed", "Sara")
mixed = ("Ali", 20, 3.45, True)

print(numbers)
print(students)
print(mixed)
````

## Single-Element Tuple

A comma is required:

```python
(10)   # integer
(10,)  # tuple
```

````python
x = (10)
y = (10,)

print(type(x))
print(type(y))
````

## Tuple Indexing and Negative Indexing

````python
students = ("Ali", "Ahmed", "Sara", "Ayesha")

print(students[0])
print(students[1])
print(students[-1])
print(students[-2])
````

## Tuple Slicing

````python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])
````

## Tuples Are Immutable

Tuple elements cannot be changed after creation.

The following would raise `TypeError`:

```python
students[1] = "Usman"
```

````python
students = ("Ali", "Ahmed", "Sara")

print("Original tuple:", students)

# Uncomment to see the TypeError:
# students[1] = "Usman"
````

## Tuple Methods — `count()` and `index()`

````python
numbers = (10, 20, 10, 30, 10)

print("10 appears:", numbers.count(10), "times")
print("20 is at index:", numbers.index(20))
````

## Looping Through Tuples

````python
students = ("Ali", "Ahmed", "Sara", "Ayesha")

for student in students:
    print(student)
````

## Tuple Unpacking

````python
student = ("Ali", 20, 3.45)

name, age, gpa = student

print("Name:", name)
print("Age:", age)
print("GPA:", gpa)
````

## AI Example — Prediction and Confidence

````python
prediction = ("Cat", 0.94)

label, confidence = prediction

print("Prediction:", label)
print("Confidence:", confidence)
````

# Part B — Sets

A set is a collection of **unique elements**.

```python
numbers = {10, 20, 30, 40}
```

Sets are useful when uniqueness and membership testing are important.

````python
numbers = {10, 20, 30, 40}
print(numbers)
print(type(numbers))
````

````python
numbers = {10, 20, 10, 30, 20, 40}
print(numbers)
````

## Empty Set

```python
{}       # empty dictionary
set()    # empty set
```

````python
empty_dictionary = {}
empty_set = set()

print(type(empty_dictionary))
print(type(empty_set))
````

## Set Membership

````python
students = {"Ali", "Ahmed", "Sara"}

print("Ali" in students)
print("Usman" in students)
````

## Adding Elements — `add()` and `update()`

````python
students = {"Ali", "Ahmed"}

students.add("Sara")
students.update(["Ayesha", "Usman"])

print(students)
````

## Removing Elements — `remove()`, `discard()`, `pop()`, `clear()`

````python
students = {"Ali", "Ahmed", "Sara"}

students.remove("Ahmed")
students.discard("Usman")

print(students)
````

````python
students = {"Ali", "Ahmed", "Sara"}

removed = students.pop()

print("Removed:", removed)
print("Remaining:", students)
````

````python
students = {"Ali", "Ahmed", "Sara"}

students.clear()

print(students)
````

# Part C — Set Operations

For:

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

we can perform union, intersection, difference, and symmetric difference.

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("A - B:", A - B)
print("B - A:", B - A)
print("Symmetric difference:", A ^ B)
````

## Set Operation Methods

```python
A.union(B)
A.intersection(B)
A.difference(B)
A.symmetric_difference(B)
```

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(A.symmetric_difference(B))
````

# Part D — Practical Data Processing

## Remove Duplicates from a List

````python
numbers = [10, 20, 10, 30, 20, 40, 30]

unique_numbers = set(numbers)

print("Original:", numbers)
print("Unique:", unique_numbers)
````

````python
students = ["Ali", "Ahmed", "Sara", "Ali", "Ayesha", "Ahmed"]

unique_students = set(students)

print("Unique students:", unique_students)
print("Number of unique students:", len(unique_students))
````

## AI Model Agreement

````python
model_a = {"Cat", "Dog", "Bird"}
model_b = {"Dog", "Horse", "Bird"}

print("Common:", model_a & model_b)
print("All:", model_a | model_b)
print("Only Model A:", model_a - model_b)
print("Only Model B:", model_b - model_a)
````

# Part E — List vs Tuple vs Set

| Feature | List | Tuple | Set |
|---|---|---|---|
| Syntax | `[]` | `()` | `{}` |
| Ordered | Yes | Yes | No positional order |
| Indexed | Yes | Yes | No |
| Mutable | Yes | No | Yes |
| Duplicates | Allowed | Allowed | Not allowed |
| Slicing | Yes | Yes | No |
| Set operations | No | No | Yes |
| Best for | Changeable collection | Fixed collection | Unique values |

## When to Use Each

### List
Use when order matters, data may change, duplicates are acceptable, and indexing/slicing are needed.

### Tuple
Use when the collection is fixed, order matters, and you want immutable data.

Examples:
```python
coordinates = (32.49, 74.52)
rgb = (255, 128, 0)
dimensions = (1920, 1080)
```

### Set
Use when values must be unique, membership testing is important, or set operations are required.
