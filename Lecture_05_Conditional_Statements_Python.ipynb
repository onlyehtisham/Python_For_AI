## 2. Decision Making

A conditional statement lets a program choose what to do based on whether a condition is `True` or `False`.

````python
age = 20

if age >= 18:
    print("You are an adult.")
````

## 3. The `if` Statement

Syntax:

```python

```

````python
marks = 80

if marks >= 50:
    print("Passed")
````

## 5. `if-else`

Use `else` when there are two possible paths.

````python
age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

### Example — Even or Odd

````python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
````

### Example — Pass or Fail

````python
marks = float(input("Enter your marks: "))

if marks >= 50:
    print("Pass")
else:
    print("Fail")
````

## 6. `if-elif-else`

Use this structure when there are more than two possible outcomes.

````python
marks = float(input("Enter your marks: "))

if marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 50:
    print("Grade D")
else:
    print("Grade F")
````

### Condition order matters

Python executes the first condition that evaluates to `True`.

````python
marks = 85

if marks >= 80:
    print("Grade A")
elif marks >= 50:
    print("Pass")
````

## 7. Logical Operators

Use `and`, `or`, and `not` to combine or reverse Boolean conditions.

````python
age = 20
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Entry denied")
````

````python
is_student = True
is_teacher = False

if is_student or is_teacher:
    print("University member")

is_raining = False
if not is_raining:
    print("You can go outside")
````

## 8. Nested Conditions

A nested condition is a conditional statement inside another conditional statement.

````python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Underage")
````

### Nested example — Student admission

````python
marks = 70
interview = 65

if marks >= 50:
    if interview >= 60:
        print("Admission successful")
    else:
        print("Interview score too low")
else:
    print("Marks requirement not met")
````

## 9. Nested Conditions vs Logical Operators

The following two approaches can express the same simple rule.

````python
# Nested conditions
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")

# Combined condition
if age >= 18 and has_id:
    print("Entry allowed")
````

## 10. AI Connection — Simple Intelligent Agent

A simple agent can receive information, evaluate conditions, and choose an action.

````python
battery = 15
obstacle = True

if battery < 20:
    print("Go to charging station")
else:
    if obstacle:
        print("Avoid obstacle")
    else:
        print("Move forward")
````

### Temperature-control agent

````python
temperature = float(input("Enter temperature: "))

if temperature >= 30:
    print("Turn on air conditioner")
elif temperature >= 20:
    print("Maintain current temperature")
else:
    print("Turn on heater")
````

### Robot battery agent

````python
battery = int(input("Enter battery percentage: "))

if battery <= 10:
    print("Critical: Go to charging station")
elif battery <= 30:
    print("Warning: Low battery")
else:
    print("Battery level is normal")
````

## 11. Practical Project — Student Grade System

````python
name = input("Enter student name: ")
marks = float(input("Enter marks: "))

if marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print(f"Student: {name}")
print(f"Marks: {marks}")
print(f"Grade: {grade}")
````

# 17. Common Beginner Mistakes

1. Forgetting the colon after a condition.
2. Incorrect indentation.
3. Using `=` instead of `==` for comparison.
4. Putting broader conditions before more specific conditions in an `if-elif-else` chain.
5. Forgetting that `input()` returns a string by default.
