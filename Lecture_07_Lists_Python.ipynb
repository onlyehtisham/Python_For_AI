## 1. Creating Lists

A list stores multiple values in one variable.

```python
students = ["Ali", "Ahmed", "Sara"]
```

Lists are ordered, changeable, and can contain multiple values.

````python
students = ["Ali", "Ahmed", "Sara", "Ayesha"]
print(students)
````

````python
marks = [78, 85, 91, 66, 72]
temperatures = [25.5, 27.3, 31.8]
results = [True, False, True]
student = ["Ali", 20, 3.45, True]

print(marks)
print(temperatures)
print(results)
print(student)
````

## 2. Empty List and `len()`

````python
students = []
students.append("Ali")
students.append("Ahmed")
print(students)
print("Length:", len(students))
````

# Part A — Indexing

Python uses zero-based indexing.

```text
Element:  Ali   Ahmed   Sara   Ayesha
Index:      0      1      2       3
```

````python
students = ["Ali", "Ahmed", "Sara", "Ayesha"]
print(students[0])
print(students[1])
print(students[2])
print(students[3])
````

````python
students = ["Ali", "Ahmed", "Sara", "Ayesha"]
print(students[-1])
print(students[-2])
print(students[-3])
print(students[-4])
````

````python
students = ["Ali", "Ahmed", "Sara"]
students[1] = "Usman"
print(students)
````

# Part B — Slicing

Syntax: `list[start:stop:step]`. The stop index is excluded.

````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[:])
````

````python
numbers = [10, 20, 30, 40, 50, 60]
print(numbers[0:6:2])
print(numbers[::-1])
````

# Part C — Adding Elements

````python
students = ["Ali", "Ahmed"]
students.append("Sara")
print(students)
````

````python
students = ["Ali", "Ahmed"]
students.extend(["Sara", "Ayesha"])
print(students)
````

````python
numbers = [1, 2]
numbers.append([3, 4])
print("append:", numbers)

numbers = [1, 2]
numbers.extend([3, 4])
print("extend:", numbers)
````

````python
students = ["Ali", "Sara"]
students.insert(1, "Ahmed")
print(students)
````

# Part D — Removing Elements

````python
students = ["Ali", "Ahmed", "Sara"]
students.remove("Ahmed")
print(students)
````

````python
students = ["Ali", "Ahmed", "Sara"]
removed = students.pop()
print("Removed:", removed)
print("Remaining:", students)
````

````python
students = ["Ali", "Ahmed", "Sara"]
removed = students.pop(1)
print("Removed:", removed)
print("Remaining:", students)
````

````python
students = ["Ali", "Ahmed", "Sara"]
students.clear()
print(students)
````

# Part E — Important List Methods

| Method | Purpose |
|---|---|
| `append()` | Add one element |
| `extend()` | Add multiple elements |
| `insert()` | Add at an index |
| `remove()` | Remove by value |
| `pop()` | Remove by index / last item |
| `clear()` | Remove all elements |
| `sort()` | Sort the list |
| `reverse()` | Reverse current order |
| `count()` | Count occurrences |
| `index()` | Find first index |

````python
numbers = [50, 10, 40, 20, 30]
numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)
````

````python
numbers = [10, 20, 30, 40]
numbers.reverse()
print(numbers)
````

````python
numbers = [10, 20, 10, 30, 10, 40]
print(numbers.count(10))
````

````python
students = ["Ali", "Ahmed", "Sara", "Ahmed"]
print(students.index("Ahmed"))
````

````python
students = ["Ali", "Ahmed", "Sara"]
print("Ali" in students)
print("Usman" in students)
````

# Part F — Looping Through Lists

````python
students = ["Ali", "Ahmed", "Sara", "Ayesha"]
for student in students:
    print(student)
````

````python
marks = [45, 78, 32, 90, 66]
for mark in marks:
    if mark >= 50:
        print(mark, "Pass")
    else:
        print(mark, "Fail")
````

````python
marks = [78, 85, 91, 66, 72]
total = 0
for mark in marks:
    total += mark
average = total / len(marks)
print("Total:", total)
print("Average:", average)
print("Highest:", max(marks))
print("Lowest:", min(marks))
````

# Part G — AI/Data Processing

````python
temperatures = [25, 28, 31, 36, 39, 42]
for temperature in temperatures:
    if temperature < 30:
        print(temperature, "Normal")
    elif temperature < 40:
        print(temperature, "Warning")
    else:
        print(temperature, "Critical")
````

````python
predictions = ["Cat", "Dog", "Cat", "Bird", "Dog"]
for prediction in predictions:
    print("AI Prediction:", prediction)

print("Cats:", predictions.count("Cat"))
print("Dogs:", predictions.count("Dog"))
print("Birds:", predictions.count("Bird"))
````

````python
sensor_data = [10, 15, -1, 20, 25, -1, 30]
valid_readings = []
for value in sensor_data:
    if value != -1:
        valid_readings.append(value)
print("Valid readings:", valid_readings)
````

# Part H — Mini Projects

````python
marks = [78, 45, 89, 32, 91, 67, 55]

total = 0
passed = 0
failed = 0

for mark in marks:
    total += mark
    if mark >= 50:
        passed += 1
    else:
        failed += 1

average = total / len(marks)
print("Students:", len(marks))
print("Total:", total)
print("Average:", average)
print("Passed:", passed)
print("Failed:", failed)
print("Highest:", max(marks))
print("Lowest:", min(marks))
````

````python
shopping = []
shopping.append("Milk")
shopping.append("Bread")
shopping.append("Eggs")

for item in shopping:
    print("-", item)

shopping.remove("Bread")
print("After removal:", shopping)
````

## Building a List from User Input

````python
students = []
for i in range(3):
    name = input("Enter student name: ")
    students.append(name)

print("Students:")
for student in students:
    print(student)
````
