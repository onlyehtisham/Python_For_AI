## 1. What is a Function?

A function is a reusable block of code that performs a specific task.

````python
# Repeating code without a function
number = 5
print(number ** 2)
number = 8
print(number ** 2)
number = 10
print(number ** 2)
````

````python
def square(number):
    return number ** 2

print(square(5))
print(square(8))
print(square(10))
````

## 2. Why Do We Need Functions?

Functions provide reusability, organization, readability, maintainability, and modularity.

## 3. Defining and Calling a Function

Use the `def` keyword. Defining a function does not execute it; calling it does.

````python
def greet():
    print("Hello, students!")

greet()
greet()
````

## 4. Functions Without Parameters

````python
def welcome():
    print("Welcome to Python Programming!")

def show_course():
    print("Python for Artificial Intelligence")

welcome()
show_course()
````

## 5. Parameters and Arguments

A **parameter** is a variable in the function definition. An **argument** is the actual value passed when the function is called.

````python
def greet(name):
    print("Hello", name)

greet("Ali")
greet("Sara")
greet("Ahmed")
````

## 6. Multiple Parameters

````python
def add(a, b):
    print(a + b)

add(10, 20)
add(100, 50)
````

````python
def student_info(name, age, department):
    print("Name:", name)
    print("Age:", age)
    print("Department:", department)

student_info("Ali", 20, "Computer Science")
````

## 7. Positional and Keyword Arguments

````python
def student_info(name, age):
    print("Name:", name)
    print("Age:", age)

# Positional arguments
student_info("Ali", 20)

# Keyword arguments
student_info(age=21, name="Sara")
````

## 8. Return Values

`return` sends a value back to the calling code. Unlike `print()`, a returned value can be stored and reused.

````python
def add(a, b):
    return a + b

result = add(10, 20)
print("Result:", result)
print("Double:", result * 2)
````

## 9. `print()` vs `return`

````python
def add_with_print(a, b):
    print(a + b)

def add_with_return(a, b):
    return a + b

add_with_print(10, 20)
result = add_with_return(10, 20)
print("Returned value:", result)
````

## 10. Returning Different Types

````python
def get_greeting(name):
    return "Hello " + name

def is_even(number):
    return number % 2 == 0

print(get_greeting("Ali"))
print(is_even(10))
print(is_even(7))
````

## 11. Returning Multiple Values

````python
def calculate(a, b):
    total = a + b
    difference = a - b
    product = a * b
    return total, difference, product

total, difference, product = calculate(20, 5)
print("Total:", total)
print("Difference:", difference)
print("Product:", product)
````

## 12. Default Parameters

````python
def greet(name="Student"):
    print("Hello", name)

greet("Ali")
greet()
````

## 13. Practical Example — Grade Calculator

````python
def calculate_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    else:
        return "F"

print(calculate_grade(85))
print(calculate_grade(72))
print(calculate_grade(45))
````

## 14. Practical Example — Average Calculator

````python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

marks = [85, 78, 92, 88]
print("Average:", calculate_average(marks))
````

## 15. Functions with Dictionaries

````python
def get_student_gpa(student):
    return student["gpa"]

student = {"name": "Ali", "age": 20, "gpa": 3.45}
print("GPA:", get_student_gpa(student))
````

````python
def calculate_average(marks):
    return sum(marks.values()) / len(marks)

marks = {"Python": 85, "AI": 91, "DSA": 88}
print("Average:", calculate_average(marks))
````
