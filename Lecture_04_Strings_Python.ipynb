## 2. What is a String?

A string is a sequence of characters. Text is represented using quotes.

```python
name = "Ali"
city = 'Sialkot'
course = "Artificial Intelligence"

print(name)
print(city)
print(course)
print(type(course))
```

### Strings can contain numbers, spaces and symbols

Anything inside quotes is treated as text.

```python
text1 = "Python 3.13"
text2 = "AI & Machine Learning"
text3 = "12345"

print(text1, type(text1))
print(text2, type(text2))
print(text3, type(text3))
```

## 4. String Indexing

Python uses zero-based indexing: the first character has index `0`.

```python
word = "PYTHON"

print(word[0])
print(word[1])
print(word[2])
print(word[5])
```

### Positive and negative indexing

`-1` means the last character, `-2` means the second-last character, and so on.

```python
word = "PYTHON"

print("First:", word[0])
print("Last:", word[-1])
print("Second last:", word[-2])
print("Third last:", word[-3])
```

## 5. String Slicing

Syntax: `string[start:end]` — start is included, end is excluded.

```python
word = "PYTHON"

print(word[0:3])
print(word[1:4])
print(word[2:5])
print(word[:3])
print(word[2:])
```

### Slicing with negative indexes and steps

```python
word = "PYTHON"

print("Last 3:", word[-3:])
print("Every 2nd character:", word[::2])
print("Reverse:", word[::-1])
```

## 6. Strings are Immutable

You cannot directly replace a character inside an existing string. Instead, create a new string.

```python
word = "Python"

# This would cause an error:
# word[0] = "J"

word = "J" + word[1:]
print(word)
```

## 7. Common String Methods

We will focus on methods that are useful for everyday programming and future AI/NLP work.

```python
text = "python programming"

print("upper():", text.upper())
print("lower():", text.lower())
print("capitalize():", text.capitalize())
print("title():", text.title())
```

### `strip()` — remove extra spaces at the beginning and end

```python
name = "   Ali Raza   "

print("Before:", repr(name))
clean_name = name.strip()
print("After:", repr(clean_name))
```

### `replace()` — replace part of a string

```python
text = "I like Java"
text = text.replace("Java", "Python")
print(text)
```

### `find()` and `count()`

```python
text = "Artificial Intelligence"

print("Position of Intelligence:", text.find("Intelligence"))
print("Position of Python:", text.find("Python"))

fruit = "banana"
print("Number of a:", fruit.count("a"))
```

### `startswith()` and `endswith()`

```python
filename = "lecture04.py"

print(filename.startswith("lecture"))
print(filename.endswith(".py"))
print(filename.endswith(".txt"))
```

## 8. `split()` and `join()`

`split()` breaks a string into pieces. `join()` combines a sequence of strings.

```python
text = "Python is easy to learn"
words = text.split()

print(words)
print("Number of words:", len(words))

sentence = " ".join(words)
print(sentence)
```

### `split()` with a delimiter

```python
data = "Ali,20,CS,3.5"
parts = data.split(",")

print(parts)
print("Name:", parts[0])
print("Age:", parts[1])
print("Department:", parts[2])
print("CGPA:", parts[3])
```

## 9. Checking String Content

These methods are useful for simple validation and text processing.

```python
print("Python".isalpha())
print("12345".isdigit())
print("ABC123".isalnum())
print("Python123".isalnum())
```

## 10. `len()`

`len()` returns the number of characters in a string.

```python
text = "Python"
print("Length:", len(text))

sentence = "Artificial Intelligence"
print("Length:", len(sentence))
```

## 11. f-Strings

f-strings provide a clean way to insert variables and expressions into text.

```python
name = "Ali"
age = 20
cgpa = 3.5

print(f"My name is {name} and I am {age} years old.")
print(f"My CGPA is {cgpa}.")
```

### Expressions inside f-strings

```python
a = 10
b = 20
marks = 450
total = 500

print(f"Sum = {a + b}")
print(f"Percentage = {(marks / total) * 100}%")
```

### Formatting numbers with f-strings

```python
price = 1234.56789
print(f"Price: {price:.2f}")
```

## 12. Practical Example — Student Profile

```python
name = input("Enter your name: ").strip()
university = input("Enter your university: ").strip()
course = input("Enter your course: ").strip()

print("\n--- Student Information ---")
print(f"Name: {name.title()}")
print(f"University: {university.title()}")
print(f"Course: {course}")
```

## 13. AI/NLP Example — Simple Text Analyzer

This is intentionally simple, but it introduces the idea of text preprocessing.

```python
text = input("Enter a sentence: ").strip()

print("\n--- Text Analysis ---")
print(f"Original text: {text}")
print(f"Uppercase: {text.upper()}")
print(f"Lowercase: {text.lower()}")
print(f"Number of characters: {len(text)}")
print(f"Number of words: {len(text.split())}")
print(f"Occurrences of 'python': {text.lower().count('python')}")
```
