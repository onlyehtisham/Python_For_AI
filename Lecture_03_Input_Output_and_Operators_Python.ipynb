# Part A — Output with `print()`

## 1. Basic `print()`

The `print()` function displays information on the screen.

````python
print("Hello, Python!")
print("Welcome to programming.")
print(100)
````

## 2. Printing Variables

````python
name = "Ali"
age = 20

print(name)
print(age)
````

## 3. Printing Multiple Values

````python
name = "Ali"
age = 20

print("Name:", name, "Age:", age)
````

## 4. `sep` Parameter

`sep` controls the separator between multiple values.

````python
print("Python", "AI", "Machine Learning", sep=" | ")
````

## 5. `end` Parameter

`end` controls what is printed after the output.

````python
print("Hello", end=" ")
print("World")
````

# Part B — Input with `input()`

## 6. Basic User Input

`input()` pauses the program and waits for the user to enter something.

````python
name = input("Enter your name: ")
print("Hello", name)
````

## 7. Important: `input()` Returns a String

Even if the user enters a number, `input()` returns text.

```python
age = input("Enter your age: ")
```

Therefore, this value is a `str`.

````python
age = input("Enter your age: ")

print("Value:", age)
print("Type:", type(age))
````

## 8. Converting Input to Integer

````python
age = int(input("Enter your age: "))

print("Age:", age)
print("Type:", type(age))
````

## 9. Converting Input to Float

````python
height = float(input("Enter your height in feet: "))

print("Height:", height)
print("Type:", type(height))
````

# Part C — Arithmetic Operators

## 10. Arithmetic Operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 3` |
| `-` | Subtraction | `10 - 3` |
| `*` | Multiplication | `10 * 3` |
| `/` | Division | `10 / 3` |
| `//` | Floor division | `10 // 3` |
| `%` | Modulus/remainder | `10 % 3` |
| `**` | Exponent | `10 ** 3` |

````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

## 11. Why `%` Is Useful

The modulus operator gives the remainder.

```python
10 % 3
```

Result:

```text
1
```

It is commonly used to check whether a number is even or odd.

````python
number = 17

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
````

# Part D — Comparison Operators

## 12. Comparison Operators

Comparison operators produce `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

````python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
````

### Important: `=` vs `==`

```python
x = 10
```

means assignment.

```python
x == 10
```

means comparison.

````python
x = 10

print(x == 10)
print(x == 20)
````

# Part E — Logical Operators

## 13. `and`

Both conditions must be true.

````python
age = 20
has_id = True

print(age >= 18 and has_id)
````

## 14. `or`

At least one condition must be true.

````python
day = "Saturday"

print(day == "Saturday" or day == "Sunday")
````

## 15. `not`

`not` reverses a Boolean value.

````python
is_raining = False

print(not is_raining)
````

# Part F — Assignment Operators

## 16. Assignment Shortcuts

Python provides convenient assignment operators:

```text
+=
-=
*=
/=
%=
**=
```

Example:

```python
score = 10
score += 5
```

This is equivalent to:

```python
score = score + 5
```

````python
score = 10

score += 5
print(score)

score -= 2
print(score)

score *= 2
print(score)

score /= 2
print(score)
````

# Part G — Practical Programs

## 17. Add Two Numbers

````python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Sum:", a + b)
````

## 18. Simple Calculator

````python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
````

## 19. Calculate Average Marks

````python
marks1 = float(input("Enter marks for subject 1: "))
marks2 = float(input("Enter marks for subject 2: "))
marks3 = float(input("Enter marks for subject 3: "))

average = (marks1 + marks2 + marks3) / 3

print("Average marks:", average)
````

## 20. Calculate Rectangle Area

````python
length = float(input("Enter length: "))
width = float(input("Enter width: "))

area = length * width

print("Area:", area)
````

## 21. Temperature Conversion

Celsius to Fahrenheit:

```text
F = (C × 9/5) + 32
```

````python
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32

print("Fahrenheit:", fahrenheit)
````

# Part H — AI Connection

````python
# Simple AI-style threshold decision
temperature = float(input("Enter temperature: "))

is_high = temperature > 35

print("Temperature:", temperature)
print("High temperature:", is_high)
````

````python
# Simple sensor calculation
sensor1 = float(input("Enter sensor 1 reading: "))
sensor2 = float(input("Enter sensor 2 reading: "))

average = (sensor1 + sensor2) / 2

print("Average sensor reading:", average)
````
