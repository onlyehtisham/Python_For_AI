## 1. What is a Loop?

A loop repeatedly executes a block of code.

Without a loop:
```python
print("Hello")
print("Hello")
print("Hello")
```

With a loop:
```python
for i in range(3):
    print("Hello")
```

Loops are useful for processing many numbers, students, customers, records, images, and AI training examples.

````python
for i in range(3):
    print("Hello")
````

## 2. Types of Loops

### `for`
Used to iterate over a sequence or range.

```python
for item in sequence:
    # code
```

### `while`
Used to repeat code while a condition is true.

```python
while condition:
    # code
```

# Part B — `range()`

`range(5)` produces `0, 1, 2, 3, 4`. The stop value is not included.

````python
for i in range(5):
    print(i)
````

````python
# range(start, stop)
for i in range(1, 6):
    print(i)
````

````python
# range(start, stop, step)
for i in range(1, 11, 2):
    print(i)
````

````python
# Counting backwards
for i in range(10, 0, -1):
    print(i)
````

## Useful `range()` examples

| Code | Values |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 5)` | 1, 2, 3, 4 |
| `range(1, 10, 2)` | 1, 3, 5, 7, 9 |
| `range(10, 0, -1)` | 10, 9, ..., 1 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 |

````python
# Countdown
for i in range(5, 0, -1):
    print(i)
print("Go!")
````

# Part C — Practical `for` Loop Examples

````python
# Numbers 1–10
for number in range(1, 11):
    print(number)
````

````python
# Even numbers
for number in range(2, 21, 2):
    print(number)
````

````python
# Multiplication table
number = 7

for i in range(1, 11):
    print(f"{number} × {i} = {number * i}")
````

````python
# Sum of 1 to 5
total = 0

for i in range(1, 6):
    total = total + i

print("Total:", total)
````

````python
# Loop through a list
students = ["Ali", "Ahmed", "Sara", "Ayesha"]

for student in students:
    print(student)
````

# Part D — Conditions Inside Loops

````python
# Print even numbers
for number in range(1, 11):
    if number % 2 == 0:
        print(number, "is even")
````

````python
# Student pass/fail
marks = [45, 78, 32, 90, 66]

for mark in marks:
    if mark >= 50:
        print(mark, "Pass")
    else:
        print(mark, "Fail")
````

````python
# AI connection — temperature classification
temperatures = [25, 31, 38, 42, 29]

for temperature in temperatures:
    if temperature > 35:
        print(temperature, "High")
    else:
        print(temperature, "Normal")
````

# Part E — `while` Loop

````python
count = 1

while count <= 5:
    print(count)
    count = count + 1
````

````python
# Countdown with while
count = 5

while count > 0:
    print(count)
    count -= 1

print("Done!")
````

## `for` vs `while`

| `for` | `while` |
|---|---|
| Iterates over a sequence/range | Runs while a condition is true |
| Often used when repetitions are known | Often used when repetitions are unknown |
| Useful for processing data | Useful for condition-based repetition |
| Usually less risk of infinite loops | Must ensure the condition can become false |

# Part F — `break`

````python
for i in range(1, 11):
    if i == 5:
        break
    print(i)
````

````python
# Search example
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        print("Number found!")
        break
````

# Part G — `continue`

````python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

````python
# Skip invalid marks
marks = [78, -5, 90, 65, -10, 88]

for mark in marks:
    if mark < 0:
        continue
    print("Valid mark:", mark)
````

````python
# AI/data processing — ignore missing values
data = [10, 20, None, 30, None, 40]

for value in data:
    if value is None:
        continue
    print("Processing:", value)
````

# Part H — Combining Loop Controls

````python
numbers = [10, -5, 20, 30, -2, 40, 50]

for number in numbers:
    if number < 0:
        continue

    if number == 40:
        break

    print(number)
````

# Part I — Nested Loops

````python
for i in range(3):
    for j in range(3):
        print(i, j)
````

````python
# Pattern
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
````

# Part K — Common Beginner Mistakes

### 1. Incorrect indentation
```python
for i in range(5):
    print(i)
```

### 2. Misunderstanding `range(5)`
It gives `0, 1, 2, 3, 4`, not `1, 2, 3, 4, 5`.

### 3. Infinite `while` loop
Always make sure the loop condition can eventually become false.

### 4. Confusing `break` and `continue`
- `break` → stop the entire loop
- `continue` → skip the current iteration
