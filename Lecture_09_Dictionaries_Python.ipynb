## 1. What is a Dictionary?

A dictionary stores data as **key-value pairs**.

```python
key → value
```

Example: `student = {"name": "Ali", "age": 20, "gpa": 3.45}`

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}

print(student)
print(type(student))
````

## 2. Creating Dictionaries

````python
student = {
    "name": "Ali", "age": 20, "department": "Computer Science",
    "semester": 5, "gpa": 3.45
}
product = {"name": "Laptop", "price": 120000, "stock": 15}
empty_dict = {}
print(student)
print(product)
print(empty_dict)
````

## 3. Keys and Values

Keys identify values. Dictionary keys should be unique.

````python
data = {"name": "Ali", "name": "Ahmed"}
print(data)  # Last value replaces the earlier one.
````

## 4. Accessing Values Using Keys

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
print(student["name"])
print(student["age"])
print(student["gpa"])
````

### Dictionary vs List

A list uses an index such as `students[0]`. A dictionary uses a meaningful key such as `student["name"]`.

## 5. Adding a New Key-Value Pair

````python
student = {"name": "Ali", "age": 20}
student["gpa"] = 3.45
student["email"] = "ali@example.com"
print(student)
````

## 6. Updating an Existing Value

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
student["gpa"] = 3.75
student["age"] = 21
print(student)
````

## 7. Deleting with `del`

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
del student["age"]
print(student)
````

## 8. Deleting with `pop()`

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
removed = student.pop("age")
print("Removed:", removed)
print(student)
````

## 9. `popitem()` and `clear()`

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
removed_item = student.popitem()
print("Removed:", removed_item)
print("Remaining:", student)
student.clear()
print("After clear:", student)
````

## 10. Important Dictionary Methods

| Method | Purpose |
|---|---|
| `keys()` | Get keys |
| `values()` | Get values |
| `items()` | Get key-value pairs |
| `get()` | Safely access a value |
| `update()` | Add/update multiple items |
| `pop()` | Remove a specific item |
| `popitem()` | Remove last item |
| `clear()` | Remove all items |
| `copy()` | Make a copy |

## 11. `keys()`, `values()`, and `items()`

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
print("Keys as list:", list(student.keys()))
````

## 12. `get()`

````python
student = {"name": "Ali", "age": 20}
print(student.get("name"))
print(student.get("email"))
print(student.get("email", "Not Available"))
````

`get()` is useful when a key may be missing. Direct access to a missing key raises `KeyError`.

## 13. `update()` and `copy()`

````python
student = {"name": "Ali", "age": 20}
student.update({"age": 21, "gpa": 3.75})
student_copy = student.copy()
print("Updated:", student)
print("Copy:", student_copy)
````

## 14. Checking Keys and Dictionary Length

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
print("name" in student)
print("email" in student)
print("email" not in student)
print("Number of items:", len(student))
````

## 15. Looping Through a Dictionary — Keys

````python
student = {"name": "Ali", "age": 20, "gpa": 3.45}
for key in student:
    print(key)
````

## 16. Looping Through Values

````python
for value in student.values():
    print(value)
````

## 17. Looping Through Keys and Values

````python
for key, value in student.items():
    print(key, ":", value)
````

## 18. Practical Example — Student Marks

````python
marks = {"Programming": 85, "Database": 78, "AI": 91, "DSA": 88}
print("AI Marks:", marks["AI"])
total = sum(marks.values())
average = total / len(marks)
print("Total:", total)
print("Average:", average)
print("Highest:", max(marks.values()))
print("Lowest:", min(marks.values()))
````

## 19. Practical Example — Product Inventory

````python
product = {"name": "Laptop", "price": 120000, "brand": "Dell", "stock": 15}
print("Product:", product["name"])
print("Price:", product["price"])
print("Stock:", product["stock"])
product["stock"] -= 1
print("Updated stock:", product["stock"])
````

# Part B — Dictionaries and AI

Dictionaries are heavily used for model predictions, confidence scores, metadata, configuration, API responses, and JSON data.

## 20. AI Example — Prediction Result

````python
prediction = {"label": "Cat", "confidence": 0.94}
print("Prediction:", prediction["label"])
print("Confidence:", prediction["confidence"])
````

## 21. AI Example — Image Classification

````python
result = {
    "image": "cat.jpg", "prediction": "Cat",
    "confidence": 0.96, "model": "ImageClassifier"
}
for key, value in result.items():
    print(key, ":", value)
````

# Part C — Nested Dictionaries

A dictionary can contain another dictionary.

````python
students = {
    "student1": {"name": "Ali", "age": 20, "gpa": 3.45},
    "student2": {"name": "Sara", "age": 21, "gpa": 3.75}
}
print(students)
````

## 22. Accessing Nested Data

````python
print(students["student1"]["name"])
print(students["student2"]["gpa"])
````

## 23. Updating Nested Data

````python
students["student1"]["gpa"] = 3.60
students["student1"]["semester"] = 5
print(students["student1"])
````

## 24. Nested Dictionary — AI Predictions

````python
predictions = {
    "image1": {"label": "Cat", "confidence": 0.95},
    "image2": {"label": "Dog", "confidence": 0.91},
    "image3": {"label": "Bird", "confidence": 0.87}
}
print(predictions["image1"]["label"])
print(predictions["image2"]["confidence"])
````

## 28. Mini Project — Student Record System

````python
student = {
    "name": "Ali", "age": 20,
    "marks": {"Python": 85, "AI": 91, "DSA": 88}
}
print("Student:", student["name"])
print("Age:", student["age"])
marks = student["marks"]
for subject, mark in marks.items():
    print(subject, ":", mark)
average = sum(marks.values()) / len(marks)
print("Average:", average)
````
