## 1. What is a Variable?

A variable is a name that refers to a value stored in a program.

Example:

```python
name = "Ali"
age = 20
```

Here:

- `name` stores `"Ali"`
- `age` stores `20`

Think of a variable as a labeled container holding a value.

````python
name = "Ali"
age = 20

print(name)
print(age)
````

## 2. Creating Variables

Python does not require a separate declaration step.

```python
name = "Ahmed"
age = 21
height = 5.8
is_student = True
```

````python
name = "Ahmed"
age = 21
height = 5.8
is_student = True

print(name)
print(age)
print(height)
print(is_student)
````

## 3. Variable Assignment

The `=` operator assigns a value to a variable.

```python
score = 85
```

This means: store the value `85` in `score`.

The `=` sign here is **assignment**, not mathematical equality.

````python
score = 85
print(score)

score = 90
print(score)
````

## 4. Multiple Assignment

Python allows multiple variables to be assigned in one line.

````python
x, y, z = 10, 20, 30

print(x)
print(y)
print(z)
````

````python
a = b = c = 100

print(a)
print(b)
print(c)
````

## 5. Variable Naming Rules

Valid examples:

```python
student_name = "Ali"
age2 = 20
_total = 500
```

Rules:

1. A name can contain letters, digits, and `_`.
2. It cannot start with a digit.
3. It cannot contain spaces.
4. Python is case-sensitive.
5. Do not use Python keywords as variable names.

Good practice:

```python
student_name = "Ali"
total_marks = 450
average_marks = 75.0
```

````python
student_name = "Ali"
total_marks = 450
average_marks = 75.0

print(student_name)
print(total_marks)
print(average_marks)
````

## 6. Python is Dynamically Typed

A variable can refer to values of different types during program execution.

````python
value = 10
print(value)

value = "Python"
print(value)

value = 3.14
print(value)
````

# Part B — Python Data Types

## 7. Integer — `int`

Integers are whole numbers without a decimal part.

Examples:

```python
10
-5
0
1000
```

````python
age = 46
temperature = -5
students = 100

print(age)
print(temperature)
print(students)
print(type(age))
````

## 8. Floating-Point Number — `float`

Floats represent numbers containing a decimal point.

Examples:

```python
3.14
5.5
-2.75
```

````python
height = 5.7
price = 1250.50
temperature = 36.5

print(height)
print(price)
print(temperature)
print(type(height))
````

## 9. String — `str`

Strings contain text and are written inside quotes.

```python
name = "Ehtisham"
course = "Python"
```

````python
name = "Ehtisham"
course = "Python"

print(name)
print(course)
print(type(name))
````

## 10. Boolean — `bool`

Boolean values have only two possibilities:

```python
True
False
```

````python
is_student = True
is_logged_in = False

print(is_student)
print(is_logged_in)
print(type(is_student))
````

## 11. Checking Data Types with `type()`

Use `type()` to find the type of a value or variable.

````python
a = 10
b = 3.14
c = "Python"
d = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))
````

## 12. Basic Data Types Summary

| Type | Example | Meaning |
|---|---|---|
| `int` | `10` | Whole number |
| `float` | `3.14` | Decimal number |
| `str` | `"Python"` | Text |
| `bool` | `True` | True/False |

# Part C — Type Conversion

## 13. Converting `str` to `int`

````python
age_text = "25"
age = int(age_text)

print(age)
print(type(age))
````

## 14. Converting `str` to `float`

````python
price_text = "125.50"
price = float(price_text)

print(price)
print(type(price))
````

## 15. Converting Numbers to Strings

````python
age = 46
age_text = str(age)

print(age_text)
print(type(age_text))
````

## 16. Converting to Boolean

````python
print(bool(1))
print(bool(0))
print(bool(""))
print(bool("Python"))
````

### Important

Common conversions:

```python
int()
float()
str()
bool()
```

Be careful: not every string can be converted to a number.

For example:

```python
int("Python")
```

will produce an error.

# Part D — Practical Examples

# Part E — AI Connection

AI programs work with many kinds of data.

Examples:

```python
model_name = "Decision Tree"
learning_rate = 0.01
epochs = 100
is_trained = False
```

Here:

- `model_name` → string
- `learning_rate` → float
- `epochs` → integer
- `is_trained` → boolean

````python
model_name = "Decision Tree"
learning_rate = 0.01
epochs = 100
is_trained = False

print("Model:", model_name)
print("Learning rate:", learning_rate)
print("Epochs:", epochs)
print("Trained:", is_trained)
````
