## 1. What is a graph?

A **graph** is a data structure that represents objects and the relationships between them.

- **Node / vertex:** an object or entity.
- **Edge:** a connection between two nodes.

A graph is often written as:

\[
G = (V, E)
\]

where `V` is the set of vertices and `E` is the set of edges.

For example, in a social network, people are nodes and friendships are edges.

````python
nodes = ["Ali", "Ahmed", "Sara", "Usman"]
edges = [("Ali", "Ahmed"), ("Ali", "Sara"),
         ("Ahmed", "Usman"), ("Sara", "Usman")]

print("Nodes:", nodes)
print("Edges:", edges)
print("Number of nodes:", len(nodes))
print("Number of edges:", len(edges))
````

## 2. Types of graphs

### Undirected graph
An undirected edge has no arrow. A connection between A and B works in both directions. Example: a mutual friendship.

### Directed graph
A directed edge has an arrow. `A → B` does not automatically mean `B → A`. Example: Ali follows Ahmed, but Ahmed may not follow Ali.

### Unweighted graph
Edges simply indicate whether a connection exists.

### Weighted graph
Each edge has a value such as distance, time, cost, or similarity.

## 3. Adjacency list

An **adjacency list** stores each node alongside its directly connected neighbors.

Consider this undirected graph:

```text
A ----- B
|       |
|       |
C ----- D
```

Its edges are A–B, A–C, B–D, and C–D. Because the graph is undirected, each connection appears in both nodes' neighbor lists.

````python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

for node, neighbors in graph.items():
    print(f"{node} -> {neighbors}")
````

### Accessing neighbors

The value associated with a node is its list of neighbors. Use `.get()` when you want a safe lookup for a node that might not exist.

````python
print("Neighbors of A:", graph["A"])

for neighbor in graph["A"]:
    print(neighbor)

print("Neighbors of unknown node:", graph.get("X", []))
````

### Adding nodes and edges safely

For an undirected graph, add each edge to both nodes' lists. The helper below also creates missing nodes and avoids duplicate edges.

````python
def add_node(graph, node):
    if node not in graph:
        graph[node] = []

def add_undirected_edge(graph, node1, node2):
    add_node(graph, node1)
    add_node(graph, node2)

    if node2 not in graph[node1]:
        graph[node1].append(node2)
    if node1 not in graph[node2]:
        graph[node2].append(node1)

social_graph = {}
add_undirected_edge(social_graph, "Ali", "Ahmed")
add_undirected_edge(social_graph, "Ali", "Sara")
add_undirected_edge(social_graph, "Ahmed", "Usman")

print(social_graph)
````

### Directed graph using an adjacency list

For a directed graph, add the neighbor only to the source node. Do not automatically add the reverse edge.

````python
def add_directed_edge(graph, source, destination):
    if source not in graph:
        graph[source] = []
    if destination not in graph:
        graph[destination] = []
    if destination not in graph[source]:
        graph[source].append(destination)

follows = {}
add_directed_edge(follows, "Ali", "Ahmed")
add_directed_edge(follows, "Ahmed", "Sara")

print(follows)
print("Who does Ali follow?", follows["Ali"])
print("Does Ahmed follow Ali?", "Ali" in follows["Ahmed"])
````

### Weighted adjacency list

For a weighted graph, store each neighbor with its weight as a tuple: `(neighbor, weight)`.

````python
weighted_graph = {
    "A": [("B", 5), ("C", 10)],
    "B": [("A", 5)],
    "C": [("A", 10)]
}

for node, neighbors in weighted_graph.items():
    print(f"{node} -> {neighbors}")
````

## 4. Adjacency matrix

An **adjacency matrix** is a two-dimensional table:
- Rows represent source nodes.
- Columns represent destination nodes.
- `1` means an edge exists; `0` means no edge exists.

For an undirected graph, the matrix is symmetric across its main diagonal.

````python
nodes = ["A", "B", "C", "D"]
matrix = [
    [0, 1, 1, 0],  # A connects to B and C
    [1, 0, 0, 1],  # B connects to A and D
    [1, 0, 0, 1],  # C connects to A and D
    [0, 1, 1, 0]   # D connects to B and C
]

print("   " + " ".join(nodes))
for node, row in zip(nodes, matrix):
    print(f"{node}  " + " ".join(map(str, row)))
````

### Reading a matrix

If `nodes = ["A", "B", "C", "D"]`, then index `0` corresponds to A and index `1` corresponds to B. Therefore, `matrix[0][1]` tells us whether A connects directly to B.

````python
print("Is A connected to B?", matrix[0][1] == 1)
print("Is A connected to D?", matrix[0][3] == 1)

node_index = {node: index for index, node in enumerate(nodes)}

def are_connected(matrix, node_index, source, destination):
    return matrix[node_index[source]][node_index[destination]] == 1

print("A-B:", are_connected(matrix, node_index, "A", "B"))
print("A-D:", are_connected(matrix, node_index, "A", "D"))
````

### Creating a matrix from an edge list

The following function builds an adjacency matrix for an **undirected, unweighted** graph. The node order determines the row and column order.

````python
def build_undirected_matrix(nodes, edges):
    index = {node: i for i, node in enumerate(nodes)}
    size = len(nodes)
    matrix = [[0] * size for _ in range(size)]

    for node1, node2 in edges:
        matrix[index[node1]][index[node2]] = 1
        matrix[index[node2]][index[node1]] = 1

    return matrix

nodes = ["A", "B", "C", "D"]
edges = [("A", "B"), ("A", "C"), ("B", "D"), ("C", "D")]
matrix = build_undirected_matrix(nodes, edges)

for node, row in zip(nodes, matrix):
    print(node, row)
````

### Weighted adjacency matrix

For a weighted matrix, store the edge weight in the relevant cell. In this example, `0` means no edge; in some applications, `None` or infinity is used instead, especially when zero could be a valid weight.

````python
weighted_matrix = [
    [0, 5, 10],
    [5, 0, 0],
    [10, 0, 0]
]
print(weighted_matrix)
````

## 5. Adjacency list vs adjacency matrix

| Feature | Adjacency list | Adjacency matrix |
|---|---|---|
| Storage | Each node's neighbors | A 2D table |
| Memory | Usually efficient for sparse graphs | Uses space for every possible pair |
| Finding all neighbors | Convenient | Scan the row |
| Checking a direct edge | May need to search a neighbor list | Direct matrix lookup |
| Best fit | Often sparse graphs | Often dense graphs or frequent edge checks |

A **sparse graph** has relatively few edges. A **dense graph** has many edges compared with the number of possible edges.

## 6. Degree and neighbors

In an undirected graph, a node's **degree** is the number of edges connected to it. A **neighbor** is a node directly connected to it.

````python
graph = {
    "A": ["C"],
    "B": ["C"],
    "C": ["A", "B", "D"],
    "D": ["C"]
}

print("Neighbors of C:", graph["C"])
print("Degree of C:", len(graph["C"]))

for node, neighbors in graph.items():
    print(f"Degree({node}) = {len(neighbors)}")
````

## 7. Practical example: social network

Let's model mutual friendships. Try adding another friendship and inspect how the graph changes.

````python
social_network = {
    "Ali": ["Ahmed", "Sara"],
    "Ahmed": ["Ali", "Usman"],
    "Sara": ["Ali", "Usman"],
    "Usman": ["Ahmed", "Sara"]
}

def show_friends(network, person):
    if person not in network:
        print(f"{person} is not in this network.")
    else:
        print(f"Friends of {person}: {network[person]}")
        print(f"Number of friends: {len(network[person])}")

show_friends(social_network, "Ali")
````

## 8. AI connection: knowledge graphs

A knowledge graph stores entities and relationships. The direction of a relationship can matter, so a directed adjacency list is often a natural choice.

````python
knowledge_graph = {
    "Python": ["AI", "Data Science"],
    "AI": ["Machine Learning"],
    "Machine Learning": ["Deep Learning"],
    "Data Science": [],
    "Deep Learning": []
}

for concept, related_concepts in knowledge_graph.items():
    print(f"{concept} -> {related_concepts}")

print("Concepts directly linked from Python:", knowledge_graph["Python"])
````

## 10. Optional: a simple Graph class

This is an extension for students who have already studied classes. The core of today's lecture remains dictionaries, lists, and matrices.

````python
class Graph:
    def __init__(self):
        self.adjacency_list = {}

    def add_node(self, node):
        add_node(self.adjacency_list, node)

    def add_edge(self, node1, node2):
        add_undirected_edge(self.adjacency_list, node1, node2)

    def display(self):
        for node, neighbors in self.adjacency_list.items():
            print(f"{node} -> {neighbors}")

g = Graph()
g.add_edge("A", "B")
g.add_edge("A", "C")
g.display()
````

## 14. Common mistakes

- **Mixing up nodes and edges:** nodes are the entities; edges are the connections.
- **Forgetting the reverse connection:** in an undirected adjacency list, add each edge to both neighbor lists.
- **Reversing a directed edge by mistake:** `A → B` does not imply `B → A`.
- **Confusing a matrix index with a node label:** keep a clear node-to-index mapping.
- **Creating duplicate edges:** check whether a neighbor is already present if duplicates are not intended.
- **Assuming every matrix uses 0 for no edge:** that is a convention; weighted graphs may use `None` or infinity instead.

## 15. Lecture summary

- A graph consists of **nodes** and **edges**.
- Edges may be directed or undirected, weighted or unweighted.
- An **adjacency list** stores each node's neighbors; Python dictionaries and lists work well.
- An **adjacency matrix** stores connections in a 2D list.
- Adjacency lists are often preferred for sparse graphs; matrices can be convenient for dense graphs and direct edge checks.
- Graphs model social networks, roads, computer networks, knowledge graphs, and recommendation systems.
